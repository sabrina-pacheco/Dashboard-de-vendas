```python
import pandas as pd
import unicodedata

df_original = pd.read_excel("alunos_sujos.xlsx")
df = df_original.copy()

def remover_acentos(texto):
    if pd.isna(texto):
        return ""
    return ''.join(
        c for c in unicodedata.normalize("NFD", str(texto))
        if unicodedata.category(c) != "Mn"
    )

def padronizar_data(valor):
    if pd.isna(valor):
        return "Não informado"

    valor = str(valor).strip()
    data = pd.to_datetime(valor, errors="coerce", dayfirst=True)

    if pd.isna(data):
        return "Não informado"

    return data.strftime("%d/%m/%Y")


# Remove linhas totalmente vazias
df = df.dropna(how="all")

# Remove linhas duplicadas
df = df.drop_duplicates()

# Padroniza nomes
df["Nome"] = df["Nome"].astype(str).str.strip()
df["Nome"] = df["Nome"].apply(remover_acentos).str.title()
df["Nome"] = df["Nome"].replace("Nan", "Não informado")

# Padroniza idade
df["Idade"] = df["Idade"].replace({
    "Dezoito": 18,
    "dezoito": 18,
    "Dezessete": 17,
    "dezessete": 17
})

df["Idade"] = pd.to_numeric(df["Idade"], errors="coerce")
df["Idade"] = df["Idade"].fillna("Não informado")

# Padroniza data de nascimento
df["Data de Nascimento"] = df["Data de Nascimento"].apply(padronizar_data)

# Padroniza gênero
df["Gênero"] = df["Gênero"].astype(str).str.strip().str.lower()
df["Gênero"] = df["Gênero"].apply(remover_acentos)

mapa_genero = {
    "m": "Masculino",
    "masc": "Masculino",
    "masc.": "Masculino",
    "masculino": "Masculino",

    "f": "Feminino",
    "fem": "Feminino",
    "fem.": "Feminino",
    "feminino": "Feminino"
}

df["Gênero"] = df["Gênero"].replace(mapa_genero)
df["Gênero"] = df["Gênero"].replace("nan", "Não informado")
df["Gênero"] = df["Gênero"].fillna("Não informado")

# Exibe tabela final
print("Tabela limpa:")
display(df)

# Exporta para HTML
df.to_html("alunos_limpos.html", index=False)

print("Arquivo HTML criado com sucesso!")
```
